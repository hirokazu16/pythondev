```python
word_list = []
char_list = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)

while True:
    input_word = input("英単語を入力してください：")
    if input_word == "":
        break
    else:
        # 文字をカウントする
        for char in input_word:
            if char in char_list:
                char_list[char] += 1
        # 単語をリストに追加
        word_list.append(input_word)
# 並べ替え 
word_list.sort()
print(word_list)

# 単語の出現回数を表示
for key, value in char_list.items():
    if value > 0:
        print(f"{key}が{value}個ありました")
```
