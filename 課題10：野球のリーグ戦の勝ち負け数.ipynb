```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
        self.total_games = win + lose
        # self.calc_win_rate()

    def calc_win_rate(self):
        if self.total_games == 0:
            # self.win_rate = 0.0
            return 0.0
        else:
            # self.win_rate = self.win / self.total_games
            return self.win / self.total_games

    def show_team_result(self,name_width,win_width,lose_width,draw_width):
        # print(f"{self.name:<{name_width}} {self.win:>{win_width}} {self.lose:>{lose_width}} {self.draw:>{draw_width}} {self.win_rate:.3f}")
        print(f"{self.name:<{name_width}} {self.win:>{win_width}} {self.lose:>{lose_width}} {self.draw:>{draw_width}} {self.calc_win_rate():.3f}")

# こうすると汎用性が無いか
# teams = [
#     Baseball_Team("Archers", 77, 64, 2),
#     Baseball_Team("Meteors", 71, 69, 3),
#     Baseball_Team("Cheetahs", 69, 68, 6),
#     Baseball_Team("Sharks", 70, 70, 3),
#     Baseball_Team("Pythons", 68, 73, 2),
#     Baseball_Team("Owls", 59, 82, 2)
# ]

data = [
    ["Archers", 77, 64, 2],
    ["Meteors", 71, 69, 3],
    ["Cheetahs", 69, 68, 6],
    ["Sharks", 70, 70, 3],
    ["Pythons", 68, 73, 2],
    ["Owls", 59, 82, 2]
]

teams = []
for d in data:
    team = Baseball_Team(d[0], d[1], d[2], d[3])
    teams.append(team)

max_name_length = 0
max_win_length = 4
max_lose_length = 4
max_draw_length = 4

for team in teams:
    if len(team.name) > max_name_length:
        max_name_length = len(team.name)
    if len(str(team.win)) > max_win_length:
        max_win_length = len(str(team.win))
    if len(str(team.lose)) > max_lose_length:
        max_lose_length = len(str(team.lose))
    if len(str(team.draw)) > max_draw_length:
        max_draw_length = len(str(team.draw))

print(
    f"{'team':<{max_name_length}} "
    f"{'win':>{max_win_length}} "
    f"{'lose':>{max_lose_length}} "
    f"{'draw':>{max_draw_length}} "
    f"{'rate':>5}"
)

for team in teams:
    team.show_team_result(max_name_length, max_win_length, max_lose_length, max_draw_length)
```
